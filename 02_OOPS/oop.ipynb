```python
class Employee:
    def __init__(self):
        print("Started executing attributes/data")
        self.id = 123
        self.salary = 60000
        self.designation = "SDE"
        print("attributes/data have been initiated")

    def travel(self,destination):
        print("This manual method was called manaually")
        print(f"The Employee is now trevelling to {destination}.")



# Obejct/instance of that class
sam = Employee()
```
